# 02 — Evaluate GTE-ModernBERT on frozen LitSearch pools

This notebook reranks exactly the first 100 documents in each saved BM25,
GRIT, and Qwen3 candidate file with both the pinned zero-shot checkpoint and
the DORIS-MAE-adapted checkpoint. Candidate membership is never changed.

```python
import os

os.environ["CUDA_DEVICE_ORDER"] = "PCI_BUS_ID"
os.environ["CUDA_VISIBLE_DEVICES"] = "5"

import sys
from pathlib import Path

def find_repo_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / "tools" / "modernbert_supervised_baseline.py").exists():
            return candidate
    raise FileNotFoundError(
        "Run this notebook from inside the cloned repository; "
        "tools/modernbert_supervised_baseline.py was not found."
    )

REPO_ROOT = find_repo_root(Path.cwd().resolve())
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
print(f"Repository root: {REPO_ROOT}")
```

```python
import gc
import json
import subprocess
import time
from pathlib import Path

import numpy as np
import torch
from datasets import load_dataset
from sentence_transformers.cross_encoder import CrossEncoder

from tools.modernbert_supervised_baseline import (
    read_jsonl,
    rerank_litsearch_rows,
    reproducibility_manifest,
    sha256_file,
    write_json,
    write_jsonl,
)

ARTIFACT_DIR = Path(os.environ.get(
    "RANKING_PRETRAINED_COMPACT_ARTIFACT_DIR",
    str(REPO_ROOT / "artifacts" / "ranking_pretrained_compact_baseline"),
))
MODEL_ID = "Alibaba-NLP/gte-reranker-modernbert-base"
MODEL_REVISION = "204bdd4d405882a9e8ca9452c223fc49f02e8451"
ADAPTED_MODEL_DIR = ARTIFACT_DIR / "models" / "gte_modernbert_dorismae_lambda_seed3407" / "final"
OUTPUT_DIR = ARTIFACT_DIR / "litsearch"
INPUT_FILES = {
    "bm25": REPO_ROOT / "LitSearch.title_abstract.bm25.jsonl",
    "grit": REPO_ROOT / "LitSearch.title_abstract.grit.jsonl",
    "qwen3": REPO_ROOT / "LitSearch.title_abstract.qwen3.jsonl",
}
DATASET_ID = "princeton-nlp/LitSearch"
DATASET_REVISION = "9573fb2"
RERANK_DEPTH = 100
MAX_LENGTH = 1024
PREDICT_BATCH_SIZE = 64
EXPECTED_QUERIES = 597

MODEL_VARIANTS = {
    "zero_shot": {
        "source": MODEL_ID,
        "revision": MODEL_REVISION,
        "output_slug": "gte_modernbert_zeroshot",
    },
    "dorismae_adapted": {
        "source": str(ADAPTED_MODEL_DIR),
        "revision": None,
        "output_slug": "gte_modernbert_dorismae",
    },
}

if not ADAPTED_MODEL_DIR.exists():
    raise FileNotFoundError(f"Run notebook 01 first; missing {ADAPTED_MODEL_DIR}")
for name, path in INPUT_FILES.items():
    if not path.exists():
        raise FileNotFoundError(f"Missing {name} candidate file: {path}")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
corpus = load_dataset(
    DATASET_ID,
    "corpus_clean",
    split="full",
    revision=DATASET_REVISION,
)
corpus_lookup = {
    int(item["corpusid"]): (
        f"Title: {item['title']}\nAbstract: {item['abstract']}"
    )
    for item in corpus
}
print(f"Corpus papers: {len(corpus_lookup):,}")
```

```python
run_records = {}
for variant_name, variant in MODEL_VARIANTS.items():
    model_kwargs = {"max_length": MAX_LENGTH}
    if variant["revision"] is not None:
        model_kwargs["revision"] = variant["revision"]
    model = CrossEncoder(variant["source"], **model_kwargs)

    def score_pairs(pairs):
        return model.predict(
            pairs,
            batch_size=PREDICT_BATCH_SIZE,
            show_progress_bar=False,
            convert_to_numpy=True,
        )

    run_records[variant_name] = {}
    for retriever, input_path in INPUT_FILES.items():
        rows = read_jsonl(input_path)
        assert len(rows) == EXPECTED_QUERIES
        assert all(len(row["retrieved"]) >= RERANK_DEPTH for row in rows)

        # Freeze the same top-100 candidate pool used by the LLM comparisons.
        rows = [
            {**row, "retrieved": [int(value) for value in row["retrieved"][:RERANK_DEPTH]]}
            for row in rows
        ]
        input_sets = [set(row["retrieved"]) for row in rows]

        if torch.cuda.is_available():
            torch.cuda.empty_cache()
            torch.cuda.reset_peak_memory_stats()
            torch.cuda.synchronize()
        started = time.perf_counter()
        reranked = rerank_litsearch_rows(rows, corpus_lookup, score_pairs)
        if torch.cuda.is_available():
            torch.cuda.synchronize()
        elapsed = time.perf_counter() - started

        assert all(
            set(row["retrieved"]) == expected
            for row, expected in zip(reranked, input_sets)
        )
        assert all(len(row["pre_reranked"]) == RERANK_DEPTH for row in reranked)
        output_path = OUTPUT_DIR / (
            f"LitSearch.title_abstract.{retriever}.{variant['output_slug']}.reranked.jsonl"
        )
        write_jsonl(output_path, reranked)
        run_records[variant_name][retriever] = {
            "input": str(input_path.resolve()),
            "input_sha256": sha256_file(input_path),
            "output": str(output_path.resolve()),
            "output_sha256": sha256_file(output_path),
            "queries": len(rows),
            "documents": len(rows) * RERANK_DEPTH,
            "elapsed_seconds": elapsed,
            "queries_per_second": len(rows) / elapsed,
            "documents_per_second": len(rows) * RERANK_DEPTH / elapsed,
            "peak_vram": (
                torch.cuda.max_memory_reserved() / (1024 ** 3)
                if torch.cuda.is_available() else 0.0
            ),
        }
        print(variant_name, retriever, json.dumps(run_records[variant_name][retriever], indent=2))

    del model
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
```

```python
# Use the same metric implementation as every existing LitSearch run.
for variant_records in run_records.values():
    for record in variant_records.values():
        command = [
            sys.executable,
            str(REPO_ROOT / "LitSearch" / "evaluate_results.py"),
            "--input",
            record["output"],
        ]
        print("Running:", " ".join(command))
        subprocess.run(command, check=True)

manifest = {
    "experiment": "gte_modernbert_zero_shot_and_dorismae_litsearch_frozen_top100",
    "model_variants": MODEL_VARIANTS,
    "dataset_id": DATASET_ID,
    "dataset_revision": DATASET_REVISION,
    "rerank_depth": RERANK_DEPTH,
    "max_length": MAX_LENGTH,
    "predict_batch_size": PREDICT_BATCH_SIZE,
    "runs": run_records,
    "environment": reproducibility_manifest(),
}
write_json(OUTPUT_DIR / "litsearch_evaluation_manifest.json", manifest)
```

The adjacent `.metrics.json` files contain aggregate values. Retain the
JSONL files because paired query-level bootstrap intervals require the
individual rankings rather than only the aggregates.
